### Missing Values

```python
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0].sort_values(ascending=False))
```

```python
missing_percentage=(df.isnull().sum() / len(df) * 100 ).sort_values(ascending=False)
missing_percentage[missing_percentage>0]
```

```python
missing_data = missing_percentage[missing_percentage>0]
plt.figure(figsize=(10,7))
plt.barh(missing_data.index, missing_data.values)
plt.title("Percentage of missing Values")
plt.xlabel("columns")
plt.ylabel("Missing percentage")
plt.tight_layout()
plt.show()
```

#####  Seconadary cause has the maximum missing values and weather condition has the minimum missing values

```python
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
    df[col] = df[col].replace("nan", np.nan)
```

```python
df["weather_condition"] = df["weather_condition"].replace({"rain": "rainy"})
for col in ["traffic_signal_present", "median_present", "mobile_phone_used"]:
    df[col] = df[col].replace({"y": "yes", "1": "yes", "true": "yes","n": "no", "0": "no", "false": "no"})
print(df["weather_condition"].unique())
print(df["traffic_signal_present"].unique())
```

```python
na_cols = ["vehicle_brand", "vehicle_model", "secondary_cause"]
for col in na_cols:
    df[col] = df[col].fillna("not applicable")
```

```python
num_missing_cols = df[numeric_cols].columns[df[numeric_cols].isnull().any()]
for col in num_missing_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_missing_cols = df[categorical_cols].columns[df[categorical_cols].isnull().any()]
for col in categorical_missing_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

### Unique Values

```python
# Negative values or impossible values
for col in numeric_cols:
    negative = (df[col] < 0).sum()
    if negative > 0:
        print(col,"->",negative,"negative values")
```

```python
df.loc[(df["driver_age"] < 0) | (df["driver_age"] > 100), "driver_age"] = np.nan
df.loc[df["num_fatalities"] < 0, "num_fatalities"] = np.nan
df.loc[(df["latitude"] < 6) | (df["latitude"] > 38), "latitude"] = np.nan
df.loc[(df["longitude"] < 68) | (df["longitude"] > 98), "longitude"] = np.nan
for col in ["driver_age", "num_fatalities", "latitude", "longitude"]:
    df[col] = df[col].fillna(df[col].median())
```

```python
df[df["estimated_speed_kmph"] > df["speed_limit_kmph"]].shape[0]
```

```python
df.loc[df["insurance_claim_filed"] != "yes", "insurance_claim_amount"] = 0
df["insurance_claim_amount"] = df["insurance_claim_amount"].fillna(df["insurance_claim_amount"].median())
```

```python
df["state"] = df["state"].str.strip()
df["district"] = df["district"].str.strip()
```

### Datetime

```python
df["year"] = df["accident_datetime"].dt.year
df["month"] = df["accident_datetime"].dt.month
df["day"] = df["accident_datetime"].dt.day
df["weekday"] = df["accident_datetime"].dt.day_name()
```

### Outliers

```python
outlier_summary = {}
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    outlier_summary[col] = count
outlier_summary
```

```python
outlier_table = pd.DataFrame(outlier_summary.items(), columns=["Column", "Outlier Count"])
outlier_table.sort_values("Outlier Count", ascending=False)
```

```python
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower=lower, upper=upper)
```

## Visualizations

```python
plt.figure(figsize=(8,5))
plt.hist(df["driver_age"], bins=20, edgecolor="black")
plt.title("Distribution of Driver Age")
plt.xlabel("Driver Age")
plt.ylabel("Number of Drivers")
plt.show()
```

##### Most drivers age who suffered from accident lies between 35 to 40 years.

```python
plt.figure(figsize=(9,5))
plt.hist(df["estimated_speed_kmph"], bins=25, edgecolor="darkblue")
plt.title("Distribution of Estimated Speed")
plt.xlabel("Speed")
plt.ylabel("Number of Accidents")
plt.show()
```

##### Most accidents occurred at moderate estimated speeds, while relatively fewer accidents occurred at very low or very high speeds

```python

plt.figure(figsize=(7,7))
plt.pie(df["accident_severity"].value_counts().values,labels=df["accident_severity"].value_counts().index,
    autopct="%1.1f%%",
    startangle=90)
plt.title("Distribution of Accident Severity")
plt.show()
```

##### The dataset contains four accident severity categories: minor, moderate, severe, and fatal. Minor accidents form the largest share, while fatal accidents represent the smallest proportion.

```python
plt.figure(figsize=(8,5))
plt.boxplot(df["num_injuries"])
plt.title("Distribution of Number of Injuries")
plt.ylabel("Number of Injuries")
plt.show()
```

##### Most accidents have small number of injuries, while some accidents have unusually high injury counts. These extreme values appear as outliers, indicating that a few accidents were much more serious than typical accidents.

```python
plt.figure(figsize=(8,5))
plt.barh(df["primary_cause"].value_counts().index, df["primary_cause"].value_counts().values)
plt.title(" Primary Causes of Accidents")
plt.xlabel("Number of Accidents")
plt.ylabel("Primary Cause")
plt.tight_layout()
plt.show()
```

##### The frequency of accidents varies considerably across different primary causes, however weather is being the major constituent in it.

```python
plt.figure(figsize=(7,7))
plt.pie(df["vehicle_type"].value_counts().values, labels=df["vehicle_type"].value_counts().index,autopct="%1.1f%%")
plt.title("Accidents by Vehicle Type",color="red")
plt.show()
```

##### Two wheelers vehicles suffered most of the accidents

```python
yearly = df["year"].value_counts().sort_index()
plt.figure(figsize=(8,5))
plt.plot(yearly.index, yearly.values,marker="o")
plt.title("Accident by Year")
plt.xlabel("Year")
plt.ylabel("Number of Accidents")
plt.grid()
plt.show()
```

##### Peaks and declines indicate that accident occurrence is not constant over time and may reflect changes in traffic conditions, reporting, or road safety.

```python
monthly = df["month"].value_counts().sort_index()
plt.figure(figsize=(8,5))
plt.plot(monthly.index,monthly.values,marker="o")
plt.title("Accidents by Month")
plt.xlabel("Month")
plt.ylabel("Number of Accidents")
plt.grid()
plt.show()
```

##### Certain months show higher accident counts, indicating possible seasonal variation in accident occurrence.

```python
df["weekday"] = df["accident_datetime"].dt.day_name()
daily = df["weekday"].value_counts()
weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
daily = daily.reindex(weekday_order)
plt.figure(figsize=(8,5))
plt.plot(daily.index, daily.values, marker="o")
plt.title("Accidents by Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Number of Accidents")
plt.grid()
plt.show()
```

##### Accidents majorly occur on weekends.

```python
road_severity = pd.crosstab(df["road_condition"], df["accident_severity"],normalize="index") * 100
road_severity.plot(kind="bar", stacked=True,figsize=(10,6))
plt.title("Road Condition vs Accident Severity")
plt.xlabel("Road Condition")
plt.ylabel("Percentage")
plt.legend(title="Severity")
plt.tight_layout()
plt.show()
```

##### Some road conditions have a relatively higher proportion of severe and fatal accidents, suggesting that road conditions may be associated with accident severity. fatal accidents least happened on dry road .

```python
alcohol_severity = pd.crosstab(df["alcohol_involved"], df["accident_severity"], normalize="index") * 100
alcohol_severity.plot(kind="bar",stacked=True, figsize=(8,5))
plt.title("Alcohol Involvement vs Accident Severity")
plt.xlabel("Alcohol Involved")
plt.ylabel("Percentage")
plt.legend(title="Severity")
plt.tight_layout()
plt.show()
```

##### Alcohol-involved accidents show a greater share of serious outcomes, indicating that alcohol involvement may be an important risk factor.

```python
vehicle_severity = pd.crosstab(df["vehicle_type"], df["accident_severity"], normalize="index") * 100
vehicle_severity.plot(kind="bar", stacked=True, figsize=(8,5))
plt.title("Vehicle Type vs Accident Severity")
plt.xlabel("Vehicle Type")
plt.ylabel("Percentage")
plt.xticks(rotation=45)
plt.legend(title="Severity")
plt.tight_layout()
plt.show()
```
