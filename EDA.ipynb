#### Datetime

```python
df["accident_datetime"] = pd.to_datetime(df["accident_datetime"])
```

```python
df["accident_datetime"].isnull().sum()
```

```python
df["year"] = df["accident_datetime"].dt.year
df["month"] = df["accident_datetime"].dt.month
df["day"] = df["accident_datetime"].dt.day
```

```python
df["accident_datetime"]
```

### Visulaization

```python
df["accident_severity"].value_counts()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["accident_severity"].value_counts().index, df["accident_severity"].value_counts().values)
plt.title("Accident Severity Distribution")
plt.xlabel("Accident Severity")
plt.ylabel("Number of Accidents")
plt.show()
```

```python
df["primary_cause"].value_counts()
```

```python
plt.figure(figsize=(8,5))
plt.bar(  df["primary_cause"].value_counts().index, df["primary_cause"].value_counts().values)
plt.title("Primary Cause Distribution") 
plt.xlabel("Primary Cause")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
