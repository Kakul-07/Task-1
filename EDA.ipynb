### Missing Values

```python
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0].sort_values(ascending=False))
```

```python
missing_percentage=(df.isnull().sum() / len(df) * 100 ).sort_values(ascending=False)
missing_percentage[missing_percentage>0]
```

```python
missing_data = missing_percentage[missing_percentage>0]
plt.figure(figsize=(10,7))
plt.bar(missing_data.index, missing_data.values)
plt.title("Percentage of missing Values")
plt.xlabel("columns")
plt.ylabel("Missing percentage")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
for col in df.columns:
    if df[col].isnull().sum() > 0:
        print(df[col].dtype)
        print("\n", col)
        print(df[col].isnull().sum())
```

```python
num_missing_cols = df[numeric_cols].columns[df[numeric_cols].isnull().any()]
for col in num_missing_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_missing_cols = df[categorical_cols].columns[df[categorical_cols].isnull().any()]
for col in categorical_missing_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

### Outliers

```python
outlier_summary = {}
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q2 = df[col].quantile(0.75)
    IQR = Q2 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q2 + 1.5 * IQR
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    outlier_summary[col] = count
outlier_summary
```

```python
outlier_table = pd.DataFrame(outlier_summary.items(),columns=["Column", "Outlier Count"])
outlier_table.sort_values("Outlier Count", ascending=False)
```

```python
df.loc[(df["driver_age"] < 0) | (df["driver_age"] > 100), "driver_age"] = np.nan
df.loc[df["num_fatalities"] < 0, "num_fatalities"] = np.nan
df.loc[(df["latitude"] < 6) | (df["latitude"] > 38), "latitude"] = np.nan
df.loc[(df["longitude"] < 68) | (df["longitude"] > 98), "longitude"] = np.nan
for col in ["driver_age","num_fatalities","latitude","longitude"]:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.loc[df["insurance_claim_filed"] != "Yes", "insurance_claim_amount"] = 0
df["insurance_claim_amount"] = df["insurance_claim_amount"].fillna(df["insurance_claim_amount"].median())
```

```python
df[["vehicle_brand","vehicle_model","secondary_cause"]]=df[["vehicle_brand","vehicle_model","secondary_cause"]].fillna("Not Applicable")
```

```python
df["state"] = df["state"].str.strip()
df["district"] = df["district"].str.strip()
```
